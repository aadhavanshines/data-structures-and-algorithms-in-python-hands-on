```python
from dsa.dynamic_array import DynamicArray

a = DynamicArray()
for i in range(10):
    a.append(i)
    print(
        f"len {len(a):>2}  capacity {a.capacity:>2}  "
        f"items copied so far {a.copies:>2}"
    )
```

Copying is O(n), but it happens so rarely that each append costs O(1) on average. Check: *n* appends copy fewer than 2*n* items.

```python
for n in (10, 1_000, 100_000):
    a = DynamicArray(range(n))
    per = a.copies / n
    print(f"{n:>7,} appends copied {a.copies:>7,} ({per:.2f} each)")
```

Popping shrinks the array at a quarter full, not at half. Shrinking at half would let one append and one pop resize every time.

```python
a = DynamicArray(range(16))
for _ in range(12):
    a.pop()
print(len(a), a.capacity)
copies = a.copies
for _ in range(1000):
    a.append(0)
    a.pop()
print("extra copies after 1,000 append/pop pairs:", a.copies - copies)
```

It behaves like a list where it matters:

```python
a = DynamicArray("abcde")
a.insert(0, "start")
a[-1] = "end"
print(a, a.pop(2), a)
```
