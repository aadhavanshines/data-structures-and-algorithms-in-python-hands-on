Each node holds a value and a link to the next node. The front is cheap to change; finding the *i*-th item means walking the links.

```python
from dsa.linked_list import LinkedList

s = LinkedList([2, 3, 4])
s.push_front(1)
s.append(5)
s
```

Draw the links, to see what the list really is:

```python
def draw(lst):
    node, parts = lst.head, []
    while node is not None:
        parts.append(f"[{node.value}]")
        node = node.next
    print(" -> ".join(parts + ["None"]))


draw(s)
s.reverse()
draw(s)
```

Two classic interview questions, both in one pass: reversing in place (above) and finding the middle with a slow and a fast pointer.

```python
for items in ([1], [1, 2, 3], [1, 2, 3, 4, 5, 6]):
    print(items, "-> middle", LinkedList(items).middle())
```

```python
s = LinkedList("abcd")
s.remove("c")
print(s.pop_front(), list(s), len(s))
```
